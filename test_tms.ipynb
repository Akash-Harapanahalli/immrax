```python
import jax
import jax.numpy as jnp
import immrax as irx
```

```python
ORDER = 2

def f (x) :
    return jnp.array([x[0]**ORDER + x[1]**ORDER - 2*x[0]**(ORDER//2)*x[1]**(ORDER//2), jnp.sin(x[0]) - x[0]])
    # return jnp.sin(x[0]) - x[0]

f_nif = jax.jit(irx.natif(f))
# f_ntm = jax.jit(lambda *args : irx.nattm(f, max_order=ORDER)(*args).interval_hull())
f_ntm = jax.jit(irx.nattm(f, max_order=ORDER))
x = irx.icentpert([0., 0.], [1., 1.])
x_tm = irx.taylor_model_from_interval(x)

f_nif(x)
f_ntm(x_tm)

nif_res, nif_times = irx.utils.run_times(100, f_nif, x)
ntm_res, ntm_times = irx.utils.run_times(100, f_ntm, x_tm)

print(f"NIF: {nif_res}, {nif_times.mean()}")
print(f"NTM: {ntm_res.interval_hull()}, {ntm_times.mean()}")
```
