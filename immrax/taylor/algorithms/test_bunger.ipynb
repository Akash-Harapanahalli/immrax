```python
import jax
import jax.numpy as jnp
import immrax as irx
```

```python
def f(t,x,w) :
    # A time decaying oscillator with unknown forcing w
    return jnp.array([
        # jnp.exp(-t) * x[1],
        # jnp.exp(-t) * (-x[0]) + w[0]
        -t * x[1],
        -t * (-x[0]) + w[0]
    ])

it0 = irx.interval([0.], [1.])
itx = irx.icentpert([1., 0.], 0.1)
itw = irx.icentpert([0.], 0.05)

# center for expansion
center = [it0.lower, itx.center, itw.center]
order = [4, 2, 1]

tm_joint = irx.taylor_model_identity([it0, itx, itw], center, order)

f_nattm = jax.jit(irx.nattm(f, structured_center=True))

tm_f, times = irx.utils.run_times(100, f_nattm, tm_joint)
print(tm_f, jnp.median(times))
```

```python
it0 = irx.interval([0.], [1.])
itx = irx.icentpert([1., 0.], [0., 0.1])
print(itx)
itw = irx.icentpert([0.], 0.05)

tm_joint_flat = irx.taylor_model_identity([it0, itx, itw], [it0.lower, itx.center, itw.center], [4, 2, 1])
tm_f_flat = f_nattm(tm_joint_flat)

print(tm_f.flat_domain, tm_f_flat.flat_domain)
print(tm_f.interval_hull(), tm_f_flat.interval_hull())
print(tm_f.remainder, tm_f_flat.remainder)
```
